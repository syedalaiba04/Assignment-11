```python
import csv

with open("Prices_Desk/teaching_prices_clean.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    data = list(reader)

rows = len(data)
columns = len(reader.fieldnames)

items = set(row["item"] for row in data)
dates = set(row["date"] for row in data)

print("Total Rows:", rows)
print("Total Columns:", columns)
print("Distinct Items:", len(items))
print("Distinct Dates:", len(dates))
```
